# DataFrame
- Series가 1차원이었다면 DataFrame은 2차원으로 확대된 버전이다.
- Excel의 SpreadSheet라고 생각하면 이해하기 쉽다.
- 2차원이기 때문에 인덱스가 row, column으로 구성되어있다
  - row: 각 개별 데이터
  - column: 각 개별의 속성의 의미
- Data Analysis, Machine Learning에서 Data를 변경하기 위해서 많이 사용된다.

## 파생 변수 만들기 연습

```python
np.nan == np.nan
```

```python
# 만약 NaN이면: 알수없음
# 19세 이상: 성인
# 19세 미만: 아이

def get_adult(age):
    if not age==age:
        return "알수없음"
    return "성인" if age >= 19 else "아이"
```

```python
get_adult(np.nan)
```

```python
def get_adult(age):
    if np.isnan(age):
        return "알수없음"
    return "성인" if age >= 19 else "아이"
```

```python
titanic_df["IsAdult"] = titanic_df["Age"].apply(get_adult)
# .apply() : 앞의 Series의 값들을 순차적으로 괄호 안의 함수에 넘겨 반환함
titanic_df
```

```python
titanic_df["Age"].apply(get_adult)
```

```python
# Fare 300 이상부터 바가지 요금, 이하는 정상요금
```

```python
def is_ripoff(fare):
    if not fare==fare:
        return "가격정보없음"
    return "바가지요금" if fare >= 300 else "정상요금"
```

```python
titanic_df["RipOff"] = titanic_df["Fare"].apply(is_ripoff)
```

```python
titanic_df
```

```python
# IsAdult RipOff 삭제
```

```python
titanic_df.drop(["IsAdult","RipOff"], inplace=True, axis=1)
titanic_df
```
