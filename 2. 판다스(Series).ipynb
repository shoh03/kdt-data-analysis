## Series
   - Pandas의 기본 객체 중 하나로, numpy의 ndarray를 기반으로 인덱싱 기능이 추가된
   1차원 배열을 나타낸다. (index x value의 형태)
   - index를 지정하지 않을 시, 기본적으로 ndarray와 같이 0-based 인덱스 생성한다
   - index를 지정할 경우 명시적으로 지정된 index를 사용한다.
   - 같은 타입의 0개 이상의 데이터를 가질 수 있다

## 시리즈(Series)의 연산
- 시리즈는 시리즈끼리 연산이 가능하다. 단, index를 기준으로 연산된다.
- 이 특징은 DataFrame에서도 동일하게 적용된다.

```python
products = {
    "치킨": 30000,
    "요아정": 50000,
    "그릭요거트": 13000,
    "커피": 6000,
    "떡볶이": 20000,
    "샌드위치": 50000
}

s2 = pd.Series(products)
s2
```

```python
products2 = {
    "치킨":30000,
    "요아정":50000,
    "국밥":13000,
    "라면":6000,
    "두쫀쿠":10000,
    "피자":30000
}
```

```python
s3 = pd.Series(products2)
```

```python
s2 + s3
# NaN과 연산하면 결과는 무조건 NaN
```

```python
s3 * 1.2
```

```python
# 가격이 3만원을 넘는 괘씸한 음식을 골라내기
price_mask = s2 > 20000
s2[price_mask]
```

```python
hobbies = {
    "승현":"칵테일, 찜질방, 노래방",
    "예나":"요리, 노래방, 영화",
    "진성":"호프, 노래방, 호프집",
    "가나디":"그릭요거트, 그림그리기, 영화",
    "지혁":"음주, 흡연, 도박",
    "시우":"볼링, 클라이밍, 보석십자수"
}

s4 = pd.Series(hobbies)
s4
```

```python
# 악의 축 취미를 가진 사용자를 mask로 제거하기 (호프, 도박, 흡연)
# contains : 각 행을 접근해서 글자들이 포함되어있으면 True, 아니면 False
evil_mask = s4.str.contains("호프|도박|흡연")
for i in s4[evil_mask].keys():
    s4.pop(i)
s4
```

```python
evil_list = ['호프', '도박', '흡연']
evil_mask = [sum(evil in hobby for evil in evil_list) == 0 for hobby in s4]
kind_member = s4[evil_mask]
kind_member
```

```python
s4[s4.str.contains("호프|도박|흡연") == 0]
```

## Series 값 삭제: .drop()
- 해당 컬럼을 삭제하고 복제본을 반환한다.

```python
s3 = pd.Series(np.arange(100,105), index=['a','b','c','d','e'])
s3
```

```python
print(s3.drop('a'))
print(s3)
# 원본 안건드림, 지울 게 없으면 오류
```

```python
print(s3.drop('a',inplace=True))
print(s3)
# 원본 건드림
```

```python
s3.drop(['b','c'])
```

## .groupby()

```python
salary_list = [
    "지혁", "희수", "승현", "우영", "희수", "시우", "슬기", "상아", "상아", "희수", "우영", "진휘", 
    "지혁", "준승", "예나", "진휘", "지호", "지혁", "희수", "승현", "시우", "준승", "우영", "예나"
]

salary_income = [
    200, 300, 400, 400, 350, 500, 300, 600, 600, 500, 600, 800,
    300, 300, 500, 500, 300, 400, 300, 800, 300, 200, 700, 500,
]
```

```python
s1 = pd.Series(data=salary_income, index=salary_list)
s1
```

```python
# .count(): 개수
s1.groupby(s1.index).count()
```

```python
# mean: 평균
s1.groupby(s1.index).mean()
```

```python
s1.groupby(s1 > 300).count()
```

```python
s1.groupby(s1 > 300).mean()
```

```python
# 600 이상 받은 사용자의 수
```

```python
s1.groupby(s1 >= 600).count()
```

```python
s1.groupby(s1 >= 600).apply(list)
```

```python
s1.groupby(s1 >= 600).apply(lambda x: list(zip(x.index,x)))
```
