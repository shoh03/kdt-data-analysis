## 맷플로립(matplotlib)
    - 파이썬의 자료(DataFrame, Series 등..)을 차트(chart)나 플롯(plot)으로 시각화(visualization)하는 모듈
    - Matplotlib은 다음과 같은 정형화된 차트나 플롯 이외에도 저수준 api를 사용한 다양한 시각화 기능을 제공한다
    - 라인 플롯(line plot)
    - 스캐터 플롯(scatter plot)
    - 컨투어 플롯(contour plot)
    - 서피스 플롯(surface plot)
    - 바 차트(bar chart)
    - 히스토그램(histogram)
    - **박스 플롯(box plot)
    - 맷플로립은 라이브러리이다.

[맷플로립 공식문서](https://matplotlib.org)

## Line plot(라인 플롯)
- 가장 간단한 플롯
- 선을 그리고 데이터가 시간, 순서등에 따라 어떻게 변하는지 보여주는 용도로 사용

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
# 윈도우 맑은 고딕 설정
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False  # 마이너스 기호 깨짐 방지
```

```python
arr = np.array([10,30,14,55])

plt.plot(arr)
```

```python
arr = np.array([
        [12,342,453],
        [63,85,523],
        [0,1,2]
    ])

plt.plot(arr)
```

```python
arr = np.array([
        [12,342,453],
        [63,85,523],
        [0,1,2]
])


plt.figure(figsize=(10,5))
for i,value in enumerate(arr):
    plt.subplot(111)
    plt.plot(arr[i])

plt.grid(True)
plt.show()
plt.title("linegraph")
```

```python
x = np.arange(0, 5, 0.1)
y = x ** 3

plt.figure(figsize=(8,4))
plt.plot(x,y)
plt.show
```

```python
test_df = pd.DataFrame(
    np.random.randn(10, 4).cumsum(axis=0),
    columns=["월", "화", "수", "목"],
    index=np.arange(0, 100, 10)
)

test_df
```

```python
plt.figure(figsize=(10,5))
plt.plot(test_df)

plt.ylabel("y축")
# y축 제목
plt.xlabel("x축")
# x축 제목
plt.legend(test_df.columns)
# 범례

plt.show()
```

## 히스토그램(histogram)
- 도수분포표를 그리는 플롯
- 가로축(계급), 세로축(도수)로 나타내는 그래프이다.

```python
# 표준편차

plt.hist(np.random.normal(0,1,size=1000))
plt.show()
```

## 산포그래프(Scatter)
- 산점도 서로 다른 두 개의 독립 변수에 대해 두 변수가 얼마나 상관관계를 가지는지 살펴보기 위해 그리는 그래프이다.

```python
# normal(평균, 표준편차, 개수)
x1 = np.random.normal(0,1,size=1000)
x2 = np.random.normal(5,1,size=1000)
y = np.random.normal(0,2,size=1000)
```

```python
plt.figure(figsize=(8,4))
plt.scatter(x1, y)
plt.scatter(x2, y)
plt.show()
```

## subplot
